```python
from langchain.agents import initialize_agent, AgentType
from langchain.tools import Tool
from langchain_groq import ChatGroq

from langchain_community.utilities import WikipediaAPIWrapper
from langchain_community.tools import DuckDuckGoSearchRun

from getpass import getpass
import os

os.environ["GROQ_API_KEY"] = getpass("Enter Groq API key: ")

llm = ChatGroq(model="llama-3.1-8b-instant")

search = DuckDuckGoSearchRun()
wiki = WikipediaAPIWrapper()

tools = [
    Tool(name="Web Search", func=search.run, description="Search internet"),
    Tool(name="Wikipedia", func=wiki.run, description="Get info"),
]

agent = initialize_agent(
    tools,
    llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

def generate_report(topic):
    prompt = f"""
    You are an expert research assistant.

    Write a COMPLETE and WELL-STRUCTURED research report on:

    {topic}

    IMPORTANT:
    - Do NOT depend too much on tools
    - Use your own knowledge + general facts
    - Write in detailed format

    FORMAT:

    -----------------------------
    COVER PAGE
    Title: {topic}
    Generated by: AI Research Agent
    -----------------------------

    1. INTRODUCTION
    (Explain topic clearly)

    2. KEY FINDINGS
    (Write 5-6 detailed bullet points)

    3. CHALLENGES
    (Write real-world problems)

    4. FUTURE SCOPE
    (What will happen in future)

    5. CONCLUSION
    (Strong summary)

    Make it detailed and professional.
    """

    return llm.invoke(prompt).content


```

```python
report1 = generate_report("Impact of AI in Healthcare")
print(report1)
```

```python
with open("report1.txt", "w") as f:
    f.write(report1)
```
